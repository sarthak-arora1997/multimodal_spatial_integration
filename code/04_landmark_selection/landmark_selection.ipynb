# Step 04: Landmark Selection in QuPath

This step is performed **manually** in [QuPath](https://qupath.github.io/) and does not require running any Python code.

## Overview

Corresponding landmark points are placed on matched channel pairs from both the DESI and Xenium modalities. These landmarks are used in the next step to compute an affine transformation for spatial registration.

## Input Images

Open the following two images in QuPath:

1. **DESI (moving image):** `data/02_channel_extraction/desi_flipped.ome.tif`  
   The horizontally flipped DESI mass spectrometry image. Use the Heme B (616.25 m/z) channel as the primary reference for vessel structures.

2. **Xenium (fixed image):** `morphology_focus_0002.ome.tif`  
   Found inside the Xenium output folder at `data/raw/xenium/output-XETG00169__0055588__55588_region_4__20250418__182706/morphology_focus/`.  
   Use the AlphaSMA/Vimentin channel (vessels) and ATP1A1/CD45/E-Cadherin channel (membranes) as references.

## Landmark Placement Workflow

1. Open both images in separate QuPath windows
2. Use the **Point** annotation tool to place landmarks on identifiable features:
   - Tissue boundaries and corners
   - Vessel branch points and intersections
   - Other distinctive morphological features visible in both modalities
3. Name each point with a consistent naming convention (e.g., `Xenium_Point_1`, `DESI_Point_1`) so point numbers can be extracted and matched
4. Place at least **10-15 landmarks** spread across the tissue for a reliable affine registration

## Exporting Landmarks

Export the landmark points from each image as **TSV** files:

1. In QuPath, select **Measure > Export measurements**
2. Export as TSV format with point coordinates (x, y) and class names
3. Save the exported files to:
   - `data/04_landmark_selection/morphology_focus_0002.ome.tif-points.tsv` (Xenium landmarks)
   - `data/04_landmark_selection/3D_DESI_F5_Pos mode_40um_Flipped.ome.tif - Image0-points.tsv` (DESI landmarks)

## Output

The `.tsv` files contain columns including `x`, `y` (pixel coordinates) and `class` (point name with number suffix). These are loaded and processed in step 05.

## Tips

- The DESI image has much lower resolution (40 um/pixel vs 0.2125 um/pixel), so landmarks on the DESI image will have much smaller coordinate values
- Focus on features that are clearly identifiable in **both** modalities
- Avoid placing landmarks in regions where one modality has poor signal
- More landmarks generally improve registration quality, but badly placed landmarks can hurt it
